# 1.1 ARTERY Overview

This notebook introduces the tutorial flow for an ARTERY-style low-latency quantum feedback path. The demo separates the offline Python analysis from the hardware interface used by the board experiment.

## Tutorial Flow

```text
S21 readout traces
  -> digital demodulation
  -> trajectory / window features
  -> state classification
  -> early confidence or BHT-style prediction
  -> feedback branch and waveform selection
```

The key idea from ARTERY is not to wait for a slow software loop. The readout trajectory is analyzed while samples arrive, and the controller can issue a feedback branch as soon as the confidence or predictor state is sufficient.

## ARTERY Blocks Used in This Tutorial

- **Trajectory Analyzer**: accumulates demodulated I/Q features over the readout trajectory.
- **Branch History Table (BHT)**: records recent trajectory patterns and their observed branch outcomes.
- **Bayesian Predictor**: converts trajectory evidence or table statistics into a probability-like confidence value.
- **Feedback Selector**: maps the predicted state or branch to a waveform that can be returned to the host or sent to a controller.
